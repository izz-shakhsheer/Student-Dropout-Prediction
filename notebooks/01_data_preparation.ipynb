# Student Dropout Prediction
## Data Preparation

This notebook prepares the dataset for all machine learning models.

The main tasks include:
- Loading and exploring the dataset.
- Transforming the target variable.
- Selecting relevant features.
- Splitting data into training and testing sets.

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../dataset/data.csv", sep=";")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
```

```python
df.head()
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
print("Missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
```

```python
print(df["Target"].value_counts())

print("\nTarget percentages:")
print(df["Target"].value_counts(normalize=True) * 100)
```

```python
df["Target_Binary"] = df["Target"].apply(
    lambda x: "Dropout" if x == "Dropout" else "Non-Dropout"
)
```

```python
print(df["Target_Binary"].value_counts())

print("\nTarget percentages:")
print(df["Target_Binary"].value_counts(normalize=True) * 100)
```

```python
second_sem_columns = [
    "Curricular units 2nd sem (credited)",
    "Curricular units 2nd sem (enrolled)",
    "Curricular units 2nd sem (evaluations)",
    "Curricular units 2nd sem (approved)",
    "Curricular units 2nd sem (grade)",
    "Curricular units 2nd sem (without evaluations)"
]

df = df.drop(columns=second_sem_columns)

print("New dataset shape:", df.shape)
```

```python
df.columns = df.columns.str.strip()

print(df.columns.tolist())
```

```python
X = df.drop(columns=["Target", "Target_Binary"])
y = df["Target_Binary"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
print("Training target distribution:")
print(y_train.value_counts(normalize=True) * 100)

print("\nTesting target distribution:")
print(y_test.value_counts(normalize=True) * 100)
```
